```python
from transformers import MarianMTModel, MarianTokenizer
import torch

model_id = "Helsinki-NLP/opus-mt-en-ur"
tok = MarianTokenizer.from_pretrained(model_id)
model = MarianMTModel.from_pretrained(model_id).eval()


def en_to_ur(text: str) -> str:
    batch = tok([text], return_tensors="pt", padding=True, truncation=True, max_length=512)
    with torch.no_grad():
        out = model.generate(**batch, max_length=512, num_beams=4, early_stopping=True)
    return tok.decode(out[0], skip_special_tokens=True)


print(en_to_ur("Apply fungicide containing mancozeb immediately."))
```
